```python
import openai
from config import OPEN_AI_KEY as api_key

openai.api_key = api_key
# import os, pprint
# for key, value in os.environ.items():
#     pprint.pprint(f"{key}: {value}")
```

```python
system_message = "You are Marv, a chatbot that reluctantly answers questions with sarcasm."
user_message = "I've recently adopted a dog. Could you suggest some dog names?"

client = openai.OpenAI(api_key=api_key)
completion = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ]
)
```

```python
completion
```

```python
print(completion.choices[0].message.content)
```

```python
user_message = "Could you explain briefly what a black hole is?"
completion = client.chat.completions.create(
    model="gpt-5.6-luna",
    messages=[
        {"role": "system", "content": system_message},
        {"role": "user", "content": user_message}
    ],
    max_completion_tokens=150,  # Not good to limit the model tokens so much.
)
```

```python
print(completion.choices[0].message.content)
```

```python
user_message = "Could you explain briefly what a black hole is?"
completion = client.chat.completions.create(
    model="gpt-5.4",  # Changed model because luna doesn't support temperature=
    messages=[
        {"role": "user", "content": user_message}
    ],
    max_completion_tokens=150,
    temperature=0
)
print(completion.choices[0].message.content)
```

```python
# Using stream
completion = client.chat.completions.create(
    model="gpt-5.4",  # Changed model because luna doesn't support temperature=
    messages=[
        {"role": "user", "content": user_message}
    ],
    max_completion_tokens=150,
    temperature=1,
    seed=365,
    stream=True
)
```

```python
for item in completion:
    print(item.choices[0].delta.content, end="")
```
