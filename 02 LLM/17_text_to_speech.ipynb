```python
import csv
import os
import tempfile
import wave

from gtts import gTTS
from IPython.display import Audio
from jiwer import wer, cer
import speech_recognition as sr
```

```python
text = "Thank you for taking the time to watch our course on speech recognition! This concludes the final lesson of this section. See you soon!"
file_path = "files/speech/Downloads/tts_output.mp3"
tts = gTTS(text, lang="en")
tts.save(file_path)
Audio(file_path)
```

```python
recognizer = sr.Recognizer()
file_path = "files/speech/Downloads/tts_output.mp3"

def transcribe_audio(file_path):
    with sr.AudioFile(file_path) as source:
        audio_data = recognizer.record(source)
        text = recognizer.recognize_google(audio_data)
        return text

def calculate_(ground_truth, transcribed_text):
    calculated = wer(ground_truth, transcribed_text), cer(ground_truth, transcribed_text)
    print(f"Word Error Rate: {calculated[0]:.4f}")
    print(f"Character Error Rate: {calculated[1]:.4f}")
```

```python
transcribed_text = transcribe_audio(file_path)
print(transcribed_text)
```
